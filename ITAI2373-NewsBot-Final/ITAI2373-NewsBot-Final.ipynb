```python
# 📊 Advanced Classification System

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.metrics import classification_report
import numpy as np


class AdvancedNewsClassifier:
    """
    Enhanced news classification with confidence scoring
    and alternative category predictions.
    """

    def __init__(self):
        # TF-IDF + Logistic Regression pipeline
        self.model = Pipeline([
            (
                "tfidf",
                TfidfVectorizer(
                    max_features=10000,
                    ngram_range=(1, 2),
                    stop_words="english"
                )
            ),
            (
                "classifier",
                LogisticRegression(
                    max_iter=1000,
                    class_weight="balanced"
                )
            )
        ])

        self.is_trained = False

    def train(self, X_train, y_train):
        """
        Train the news classification model.
        """

        self.model.fit(X_train, y_train)
        self.is_trained = True

        print("✅ Advanced classifier trained successfully!")

    def predict_with_confidence(self, article_text):
        """
        Predict the main news category and return confidence scores.
        """

        if not self.is_trained:
            return {
                "status": "error",
                "message": "Classifier must be trained first."
            }

        # Predict class probabilities
        probabilities = self.model.predict_proba([article_text])[0]
        classes = self.model.classes_

        # Sort categories by confidence
        sorted_indices = np.argsort(probabilities)[::-1]

        primary_index = sorted_indices[0]

        primary_category = classes[primary_index]
        primary_confidence = probabilities[primary_index]

        alternatives = []

        for index in sorted_indices[1:4]:
            alternatives.append({
                "category": classes[index],
                "confidence": round(float(probabilities[index]), 4)
            })

        return {
            "primary_category": primary_category,
            "confidence": round(float(primary_confidence), 4),
            "alternative_categories": alternatives
        }

    def explain_prediction(self, article_text, top_n=10):
        """
        Explain which words influenced the classification decision.
        """

        if not self.is_trained:
            return {
                "status": "error",
                "message": "Classifier must be trained first."
            }

        tfidf = self.model.named_steps["tfidf"]
        classifier = self.model.named_steps["classifier"]

        vectorized_text = tfidf.transform([article_text])

        predicted_class = self.model.predict([article_text])[0]
        class_index = list(classifier.classes_).index(predicted_class)

        feature_names = np.array(tfidf.get_feature_names_out())

        feature_values = vectorized_text.toarray()[0]

        coefficients = classifier.coef_[class_index]

        importance = feature_values * coefficients

        top_indices = np.argsort(importance)[::-1][:top_n]

        key_phrases = [
            feature_names[i]
            for i in top_indices
            if importance[i] > 0
        ]

        return {
            "predicted_category": predicted_class,
            "important_words": key_phrases
        }


# Initialize classifier
classifier = AdvancedNewsClassifier()

print("📊 Advanced classification system ready!")
```

```python
# 🔍 Topic Modeling and Discovery

from sklearn.feature_extraction.text import CountVectorizer
from sklearn.decomposition import LatentDirichletAllocation
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


class TopicDiscoveryEngine:
    """
    Advanced topic modeling for discovering themes and trends.
    Uses LDA to identify hidden topics in news articles.
    """

    def __init__(self, n_topics=10, method="lda"):
        self.n_topics = n_topics
        self.method = method

        # Convert text into word-count features
        self.vectorizer = CountVectorizer(
            max_df=0.95,
            min_df=2,
            stop_words="english",
            max_features=5000
        )

        # LDA topic model
        self.model = LatentDirichletAllocation(
            n_components=n_topics,
            random_state=42,
            learning_method="batch"
        )

        self.feature_names = None
        self.is_fitted = False


    def fit_topics(self, documents):
        """
        Discover topics from a collection of documents.
        """

        # Convert documents to document-term matrix
        document_matrix = self.vectorizer.fit_transform(documents)

        # Train LDA model
        self.model.fit(document_matrix)

        self.feature_names = self.vectorizer.get_feature_names_out()
        self.is_fitted = True

        print(f"✅ Discovered {self.n_topics} topics.")

        return self.get_top_words()


    def get_top_words(self, top_n=10):
        """
        Return the most important words for each topic.
        """

        if not self.is_fitted:
            return "Topic model has not been trained yet."

        topics = {}

        for topic_index, topic in enumerate(self.model.components_):

            top_indices = topic.argsort()[-top_n:][::-1]

            words = [
                self.feature_names[index]
                for index in top_indices
            ]

            topics[f"Topic {topic_index + 1}"] = words

        return topics


    def get_article_topics(self, article_text):
        """
        Get the topic distribution for one article.
        """

        if not self.is_fitted:
            return {
                "status": "error",
                "message": "Train the topic model first."
            }

        article_vector = self.vectorizer.transform([article_text])

        topic_distribution = self.model.transform(article_vector)[0]

        results = []

        for i, score in enumerate(topic_distribution):

            results.append({
                "topic": f"Topic {i + 1}",
                "score": round(float(score), 4)
            })

        results = sorted(
            results,
            key=lambda x: x["score"],
            reverse=True
        )

        return results


    def track_topic_trends(self, articles_with_dates):
        """
        Analyze how topics change over time.

        Expected columns:
        - text
        - date
        """

        if not self.is_fitted:
            return "Train the topic model first."

        data = articles_with_dates.copy()

        data["date"] = pd.to_datetime(data["date"])

        document_matrix = self.vectorizer.transform(data["text"])

        topic_scores = self.model.transform(document_matrix)

        # Assign dominant topic
        data["dominant_topic"] = np.argmax(topic_scores, axis=1) + 1

        # Group by month and topic
        data["month"] = data["date"].dt.to_period("M").astype(str)

        trends = (
            data.groupby(["month", "dominant_topic"])
            .size()
            .reset_index(name="article_count")
        )

        return trends


    def visualize_topics(self):
        """
        Create a simple visualization of top words per topic.
        """

        if not self.is_fitted:
            print("Train the topic model first.")
            return

        topics = self.get_top_words(top_n=10)

        for topic_name, words in topics.items():

            print(f"\n{topic_name}")
            print(", ".join(words))


# Initialize topic engine
topic_engine = TopicDiscoveryEngine(
    n_topics=5,
    method="lda"
)

print("🔍 Topic discovery engine ready!")
```

```python
# 🎭 Advanced Sentiment Analysis

from transformers import pipeline
import pandas as pd
import numpy as np


class SentimentEvolutionTracker:
    """
    Advanced sentiment analysis with temporal tracking.
    Uses a pretrained transformer model for sentiment analysis.
    """

    def __init__(self):
        # Pretrained sentiment analysis model
        self.sentiment_model = pipeline(
            "sentiment-analysis"
        )

        self.is_ready = True


    def analyze_sentiment(self, article_text):
        """
        Perform sentiment analysis on a news article.

        Returns:
        - Overall sentiment
        - Confidence score
        - Basic emotional interpretation
        - Key sentiment phrases
        """

        if not article_text:
            return {
                "status": "error",
                "message": "No article text provided."
            }

        # Transformer models may have input length limits
        shortened_text = article_text[:1500]

        result = self.sentiment_model(shortened_text)[0]

        label = result["label"].lower()
        confidence = float(result["score"])

        # Convert common model labels
        if "positive" in label:
            overall_sentiment = "positive"
        elif "negative" in label:
            overall_sentiment = "negative"
        else:
            overall_sentiment = "neutral"

        # Simple emotional interpretation
        emotions = {
            "joy": 0.0,
            "anger": 0.0,
            "fear": 0.0,
            "sadness": 0.0
        }

        if overall_sentiment == "positive":
            emotions["joy"] = confidence

        elif overall_sentiment == "negative":
            emotions["anger"] = confidence * 0.5
            emotions["fear"] = confidence * 0.3
            emotions["sadness"] = confidence * 0.2

        return {
            "overall_sentiment": overall_sentiment,
            "confidence": round(confidence, 4),
            "emotions": emotions
        }


    def track_sentiment_over_time(self, articles_with_dates):
        """
        Analyze sentiment trends over time.

        Expected DataFrame columns:
        - text
        - date
        """

        data = articles_with_dates.copy()

        data["date"] = pd.to_datetime(data["date"])

        sentiment_scores = []
        sentiment_labels = []

        for article in data["text"]:

            result = self.analyze_sentiment(article)

            sentiment_labels.append(
                result["overall_sentiment"]
            )

            # Convert sentiment into numerical score
            if result["overall_sentiment"] == "positive":
                score = result["confidence"]

            elif result["overall_sentiment"] == "negative":
                score = -result["confidence"]

            else:
                score = 0

            sentiment_scores.append(score)

        data["sentiment"] = sentiment_labels
        data["sentiment_score"] = sentiment_scores

        # Create monthly time periods
        data["month"] = (
            data["date"]
            .dt.to_period("M")
            .astype(str)
        )

        # Average sentiment by month
        sentiment_timeline = (
            data.groupby("month")["sentiment_score"]
            .mean()
            .reset_index()
        )

        return sentiment_timeline


    def detect_sentiment_anomalies(self, sentiment_timeline):
        """
        Identify unusual sentiment changes.

        Uses standard deviation to detect large changes
        in sentiment over time.
        """

        timeline = sentiment_timeline.copy()

        mean_score = timeline["sentiment_score"].mean()

        std_score = timeline["sentiment_score"].std()

        if std_score == 0 or np.isnan(std_score):
            timeline["anomaly"] = False
            return timeline

        timeline["z_score"] = (
            timeline["sentiment_score"] - mean_score
        ) / std_score

        # Mark unusually large sentiment changes
        timeline["anomaly"] = (
            timeline["z_score"].abs() > 2
        )

        return timeline


# Initialize sentiment tracker
sentiment_tracker = SentimentEvolutionTracker()

print("🎭 Sentiment evolution tracker ready!")
```

```python
# 🕸️ Entity Relationship Mapping

import spacy
import networkx as nx
from collections import defaultdict


class EntityRelationshipMapper:
    """
    Advanced NER with relationship extraction
    and knowledge graph construction.
    """

    def __init__(self):
        # Load spaCy English NLP model
        try:
            self.nlp = spacy.load("en_core_web_sm")
        except:
            print("⚠️ spaCy model not found.")
            print("Run: !python -m spacy download en_core_web_sm")
            self.nlp = None

        # Knowledge graph
        self.graph = nx.Graph()

        # Store extracted entities
        self.entity_store = defaultdict(list)


    def extract_entities(self, article_text):
        """
        Extract and classify named entities.

        Identifies:
        - People
        - Organizations
        - Locations
        - Events
        - Products
        - Dates
        """

        if self.nlp is None:
            return {
                "status": "error",
                "message": "spaCy model is not loaded."
            }

        if not article_text:
            return {
                "status": "error",
                "message": "No article text provided."
            }

        doc = self.nlp(article_text)

        entities = []

        for ent in doc.ents:

            entity_info = {
                "text": ent.text,
                "label": ent.label_,
                "description": spacy.explain(ent.label_)
            }

            entities.append(entity_info)

            self.entity_store[ent.label_].append(ent.text)

        return entities


    def extract_relationships(self, article_text):
        """
        Extract simple relationships between entities.

        Relationships are based on entities appearing
        together in the same sentence.
        """

        if self.nlp is None:
            return []

        doc = self.nlp(article_text)

        relationships = []

        for sentence in doc.sents:

            sentence_entities = [
                ent for ent in sentence.ents
                if ent.label_ in [
                    "PERSON",
                    "ORG",
                    "GPE",
                    "LOC",
                    "EVENT",
                    "PRODUCT"
                ]
            ]

            # Create relationships between entity pairs
            for i in range(len(sentence_entities)):

                for j in range(i + 1, len(sentence_entities)):

                    entity1 = sentence_entities[i]
                    entity2 = sentence_entities[j]

                    relationship = {
                        "source": entity1.text,
                        "source_type": entity1.label_,
                        "target": entity2.text,
                        "target_type": entity2.label_,
                        "relationship": "mentioned_with",
                        "context": sentence.text.strip()
                    }

                    relationships.append(relationship)

        return relationships


    def build_knowledge_graph(self, articles):
        """
        Build a knowledge graph from multiple articles.

        Nodes represent entities.
        Edges represent relationships.
        """

        self.graph.clear()

        for article in articles:

            relationships = self.extract_relationships(article)

            for relation in relationships:

                source = relation["source"]
                target = relation["target"]

                self.graph.add_node(
                    source,
                    entity_type=relation["source_type"]
                )

                self.graph.add_node(
                    target,
                    entity_type=relation["target_type"]
                )

                # Increase connection weight if repeated
                if self.graph.has_edge(source, target):

                    self.graph[source][target]["weight"] += 1

                else:

                    self.graph.add_edge(
                        source,
                        target,
                        relationship=relation["relationship"],
                        weight=1
                    )

        return {
            "entities": self.graph.number_of_nodes(),
            "relationships": self.graph.number_of_edges()
        }


    def find_entity_connections(self, entity1, entity2):
        """
        Find connections between two entities.
        """

        if entity1 not in self.graph:
            return {
                "status": "error",
                "message": f"{entity1} is not in the knowledge graph."
            }

        if entity2 not in self.graph:
            return {
                "status": "error",
                "message": f"{entity2} is not in the knowledge graph."
            }

        try:

            path = nx.shortest_path(
                self.graph,
                source=entity1,
                target=entity2
            )

            return {
                "status": "connected",
                "path": path,
                "degrees_of_separation": len(path) - 1
            }

        except nx.NetworkXNoPath:

            return {
                "status": "not connected",
                "message": (
                    f"No relationship was found between "
                    f"{entity1} and {entity2}."
                )
            }


    def get_top_entities(self, top_n=10):
        """
        Identify the most connected entities
        in the knowledge graph.
        """

        if self.graph.number_of_nodes() == 0:
            return []

        degree_scores = nx.degree_centrality(self.graph)

        ranked_entities = sorted(
            degree_scores.items(),
            key=lambda x: x[1],
            reverse=True
        )

        return ranked_entities[:top_n]


# Initialize entity mapper
entity_mapper = EntityRelationshipMapper()

print("🕸️ Entity relationship mapper ready!")
```

```python
# 📝 Intelligent Text Summarization - FIXED VERSION

from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
import torch
import re


class IntelligentSummarizer:
    """
    Intelligent text summarization using a pretrained BART model.
    """

    def __init__(self):

        self.model_name = "facebook/bart-large-cnn"

        print("Loading summarization model...")

        self.tokenizer = AutoTokenizer.from_pretrained(
            self.model_name
        )

        self.model = AutoModelForSeq2SeqLM.from_pretrained(
            self.model_name
        )

        self.device = torch.device(
            "cuda" if torch.cuda.is_available() else "cpu"
        )

        self.model.to(self.device)

        print("✅ Summarization model loaded!")


    def summarize_article(
        self,
        article_text,
        summary_type="balanced"
    ):
        """
        Generate a summary for one article.
        """

        if not article_text:
            return {
                "status": "error",
                "message": "No article text provided."
            }

        length_settings = {
            "brief": {
                "max_length": 60,
                "min_length": 20
            },

            "balanced": {
                "max_length": 130,
                "min_length": 40
            },

            "detailed": {
                "max_length": 220,
                "min_length": 80
            }
        }

        settings = length_settings.get(
            summary_type,
            length_settings["balanced"]
        )

        # Tokenize article
        inputs = self.tokenizer(
            article_text,
            return_tensors="pt",
            max_length=1024,
            truncation=True
        )

        inputs = {
            key: value.to(self.device)
            for key, value in inputs.items()
        }

        # Generate summary
        with torch.no_grad():

            summary_ids = self.model.generate(
                **inputs,
                max_length=settings["max_length"],
                min_length=settings["min_length"],
                num_beams=4,
                early_stopping=True
            )

        summary = self.tokenizer.decode(
            summary_ids[0],
            skip_special_tokens=True
        )

        return {
            "summary_type": summary_type,
            "summary": summary,
            "original_word_count": len(
                article_text.split()
            ),
            "summary_word_count": len(
                summary.split()
            )
        }


    def summarize_multiple_articles(
        self,
        articles,
        focus_topic=None
    ):

        if not articles:
            return {
                "status": "error",
                "message": "No articles provided."
            }

        combined_text = " ".join(
            article
            for article in articles
            if isinstance(article, str)
        )

        if focus_topic:

            combined_text = (
                f"Focus on {focus_topic}. "
                + combined_text
            )

        result = self.summarize_article(
            combined_text,
            summary_type="detailed"
        )

        return {
            "article_count": len(articles),
            "focus_topic": focus_topic,
            "summary": result["summary"]
        }


    def generate_headlines(self, article_text):

        result = self.summarize_article(
            article_text,
            summary_type="brief"
        )

        summary = result["summary"]

        first_sentence = re.split(
            r"[.!?]",
            summary
        )[0].strip()

        return {
            "informative":
                first_sentence[:90],

            "engaging":
                f"Why This Story Matters: "
                f"{first_sentence[:65]}",

            "seo_optimized":
                f"{first_sentence[:65]} | Latest News",

            "social_media":
                f"Breaking: {first_sentence[:70]}"
        }


    def assess_summary_quality(
        self,
        original_text,
        summary
    ):

        original_words = set(
            re.findall(
                r"\b\w+\b",
                original_text.lower()
            )
        )

        summary_words = set(
            re.findall(
                r"\b\w+\b",
                summary.lower()
            )
        )

        overlap = original_words.intersection(
            summary_words
        )

        coverage = (
            len(overlap) / len(summary_words)
            if summary_words else 0
        )

        compression_ratio = (
            len(summary.split())
            / len(original_text.split())
            if original_text.split()
            else 0
        )

        return {
            "compression_ratio":
                round(compression_ratio, 3),

            "information_coverage":
                round(coverage, 3),

            "summary_word_count":
                len(summary.split())
        }


# Initialize summarizer
summarizer = IntelligentSummarizer()

print("📝 Intelligent summarizer ready!")
```

```python
# 🔍 Semantic Search and Similarity

from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.cluster import KMeans
import numpy as np
import pandas as pd


class SemanticSearchEngine:
    """
    Semantic search using sentence embeddings
    and cosine similarity.
    """

    def __init__(self):
        # Small, fast sentence-transformer model
        self.model_name = "all-MiniLM-L6-v2"

        print("Loading semantic embedding model...")

        self.model = SentenceTransformer(
            self.model_name
        )

        self.documents = []
        self.document_embeddings = None

        print("✅ Semantic model loaded!")


    def encode_documents(self, documents):
        """
        Convert documents into semantic embeddings.
        """

        if not documents:
            return {
                "status": "error",
                "message": "No documents provided."
            }

        self.documents = documents

        self.document_embeddings = self.model.encode(
            documents,
            convert_to_numpy=True,
            show_progress_bar=False
        )

        return {
            "status": "success",
            "document_count": len(documents),
            "embedding_shape": self.document_embeddings.shape
        }


    def find_similar_articles(
        self,
        query_article,
        top_k=5
    ):
        """
        Find articles most similar to a query article.
        """

        if self.document_embeddings is None:
            return {
                "status": "error",
                "message": "Encode documents first."
            }

        # Encode the query article
        query_embedding = self.model.encode(
            [query_article],
            convert_to_numpy=True
        )

        # Calculate cosine similarity
        similarities = cosine_similarity(
            query_embedding,
            self.document_embeddings
        )[0]

        # Get highest similarity scores
        top_indices = np.argsort(
            similarities
        )[::-1][:top_k]

        results = []

        for index in top_indices:
            results.append({
                "article_index": int(index),
                "article": self.documents[index],
                "similarity_score": round(
                    float(similarities[index]),
                    4
                )
            })

        return results


    def semantic_search(
        self,
        query_text,
        article_database,
        top_k=5
    ):
        """
        Search articles using a natural language query.
        """

        if not article_database:
            return {
                "status": "error",
                "message": "Article database is empty."
            }

        # Encode articles
        article_embeddings = self.model.encode(
            article_database,
            convert_to_numpy=True,
            show_progress_bar=False
        )

        # Encode query
        query_embedding = self.model.encode(
            [query_text],
            convert_to_numpy=True
        )

        # Compute similarity
        similarities = cosine_similarity(
            query_embedding,
            article_embeddings
        )[0]

        top_indices = np.argsort(
            similarities
        )[::-1][:top_k]

        search_results = []

        for index in top_indices:
            search_results.append({
                "rank": len(search_results) + 1,
                "article": article_database[index],
                "similarity_score": round(
                    float(similarities[index]),
                    4
                )
            })

        return search_results


    def cluster_similar_content(
        self,
        articles,
        n_clusters=3
    ):
        """
        Group semantically similar articles
        using K-Means clustering.
        """

        if not articles:
            return {
                "status": "error",
                "message": "No articles provided."
            }

        # Do not allow more clusters than articles
        n_clusters = min(
            n_clusters,
            len(articles)
        )

        embeddings = self.model.encode(
            articles,
            convert_to_numpy=True,
            show_progress_bar=False
        )

        clustering_model = KMeans(
            n_clusters=n_clusters,
            random_state=42,
            n_init=10
        )

        labels = clustering_model.fit_predict(
            embeddings
        )

        results = pd.DataFrame({
            "article": articles,
            "cluster": labels
        })

        return results


# Initialize semantic search engine
search_engine = SemanticSearchEngine()

print("🔍 Semantic search engine ready!")
```

```python
# 💡 Content Enhancement and Insights

import re
from collections import Counter
import pandas as pd


class ContentEnhancer:
    """
    Content analysis and automatic insight generation.
    Provides context, trends, gaps, and basic fact-check support.
    """

    def __init__(self):
        # Important terms that may indicate missing perspectives
        self.perspective_keywords = {
            "economic": [
                "cost", "market", "business", "jobs",
                "economic", "investment"
            ],
            "social": [
                "community", "public", "society",
                "people", "education", "health"
            ],
            "technology": [
                "technology", "AI", "software",
                "cybersecurity", "data"
            ],
            "government": [
                "government", "policy", "law",
                "regulation", "official"
            ],
            "environment": [
                "climate", "environment",
                "energy", "pollution"
            ]
        }


    def enhance_article(self, article_text):
        """
        Add useful context and basic insights to an article.
        """

        if not article_text:
            return {
                "status": "error",
                "message": "No article text provided."
            }

        words = re.findall(
            r"\b[a-zA-Z]{4,}\b",
            article_text.lower()
        )

        word_counts = Counter(words)

        # Remove very common words
        common_words = {
            "this", "that", "with", "from",
            "have", "will", "their", "about",
            "were", "been", "they", "said"
        }

        important_terms = [
            word for word, count in word_counts.most_common(15)
            if word not in common_words
        ]

        # Detect basic themes
        detected_themes = []

        lower_text = article_text.lower()

        for theme, keywords in self.perspective_keywords.items():

            if any(
                keyword.lower() in lower_text
                for keyword in keywords
            ):
                detected_themes.append(theme)

        return {
            "word_count": len(article_text.split()),
            "important_terms": important_terms[:10],
            "detected_themes": detected_themes,
            "context_suggestion":
                "Review related articles and trusted sources "
                "to add historical and factual context."
        }


    def generate_insights(self, articles):
        """
        Generate high-level insights from multiple articles.
        """

        if not articles:
            return {
                "status": "error",
                "message": "No articles provided."
            }

        combined_text = " ".join(
            article for article in articles
            if isinstance(article, str)
        )

        words = re.findall(
            r"\b[a-zA-Z]{4,}\b",
            combined_text.lower()
        )

        common_words = {
            "this", "that", "with", "from",
            "have", "will", "their", "about",
            "were", "been", "they", "said"
        }

        word_counts = Counter(
            word for word in words
            if word not in common_words
        )

        trending_terms = word_counts.most_common(10)

        # Detect themes across article collection
        theme_counts = {}

        for theme, keywords in self.perspective_keywords.items():

            count = 0

            for article in articles:

                text = article.lower()

                if any(
                    keyword.lower() in text
                    for keyword in keywords
                ):
                    count += 1

            theme_counts[theme] = count

        return {
            "article_count": len(articles),
            "trending_terms": trending_terms,
            "theme_frequency": theme_counts,
            "insight":
                "Frequently repeated terms and themes may "
                "represent important trends in the collection."
        }


    def detect_information_gaps(self, articles, topic):
        """
        Identify perspectives that may be missing
        from coverage of a topic.
        """

        if not articles:
            return {
                "status": "error",
                "message": "No articles provided."
            }

        combined_text = " ".join(
            articles
        ).lower()

        missing_perspectives = []
        covered_perspectives = []

        for perspective, keywords in self.perspective_keywords.items():

            found = any(
                keyword.lower() in combined_text
                for keyword in keywords
            )

            if found:
                covered_perspectives.append(
                    perspective
                )
            else:
                missing_perspectives.append(
                    perspective
                )

        follow_up_questions = [
            f"What is the {perspective} impact of {topic}?"
            for perspective in missing_perspectives
        ]

        return {
            "topic": topic,
            "covered_perspectives":
                covered_perspectives,
            "missing_perspectives":
                missing_perspectives,
            "suggested_questions":
                follow_up_questions
        }


    def cross_reference_facts(self, article_text):
        """
        Identify statements that may need external verification.

        Note:
        This does not confirm whether a claim is true.
        It flags factual-looking statements for review.
        """

        if not article_text:
            return {
                "status": "error",
                "message": "No article text provided."
            }

        sentences = re.split(
            r"(?<=[.!?])\s+",
            article_text
        )

        claims_to_check = []

        for sentence in sentences:

            # Look for numbers, percentages,
            # dates, or strong factual language
            if (
                re.search(r"\d", sentence)
                or "%" in sentence
                or "according to" in sentence.lower()
                or "reported" in sentence.lower()
                or "study" in sentence.lower()
            ):
                claims_to_check.append(
                    sentence.strip()
                )

        return {
            "claims_detected": len(
                claims_to_check
            ),
            "claims_to_verify":
                claims_to_check,
            "verification_note":
                "These statements should be checked "
                "against reliable external sources."
        }


# Initialize content enhancer
enhancer = ContentEnhancer()

print("💡 Content enhancer ready!")
```

```python
# 🌐 Language Detection and Processing

!pip install langdetect
from langdetect import detect_langs, DetectorFactory
from transformers import pipeline
from collections import Counter
import re

# Make language detection results reproducible
DetectorFactory.seed = 42


class MultilingualProcessor:
    """
    Multilingual processing with language detection,
    translation, cross-lingual comparison,
    and basic cultural context analysis.
    """

    def __init__(self):
        # Simple language name mapping
        self.language_names = {
            "en": "English",
            "es": "Spanish",
            "fr": "French",
            "de": "German",
            "it": "Italian",
            "pt": "Portuguese",
            "th": "Thai",
            "zh-cn": "Chinese",
            "ja": "Japanese",
            "ko": "Korean"
        }

        # Lazy-load translation model only when needed
        self.translation_models = {}

        # Basic cultural/regional reference keywords
        self.cultural_keywords = {
            "United States": [
                "congress", "white house", "federal",
                "state", "president"
            ],
            "Thailand": [
                "bangkok", "thai", "baht",
                "songkran", "province"
            ],
            "Europe": [
                "european union", "eu",
                "brussels", "euro"
            ],
            "Asia": [
                "asean", "asia",
                "regional cooperation"
            ]
        }


    def detect_language(self, text):
        """
        Detect the most likely language
        with confidence scoring.
        """

        if not text or len(text.strip()) < 3:
            return {
                "status": "error",
                "message": "Text is too short for reliable detection."
            }

        try:
            results = detect_langs(text)

            primary = results[0]

            language_code = primary.lang
            confidence = float(primary.prob)

            alternatives = []

            for result in results[1:3]:
                alternatives.append({
                    "language": result.lang,
                    "confidence": round(
                        float(result.prob),
                        4
                    )
                })

            return {
                "language_code": language_code,
                "language_name": self.language_names.get(
                    language_code,
                    language_code
                ),
                "confidence": round(
                    confidence,
                    4
                ),
                "alternative_languages": alternatives
            }

        except Exception as e:
            return {
                "status": "error",
                "message": str(e)
            }


    def translate_text(
        self,
        text,
        target_language="en"
    ):
        """
        Translate text into a target language.

        Currently supports selected language pairs
        using Hugging Face translation models.
        """

        if not text:
            return {
                "status": "error",
                "message": "No text provided."
            }

        detected = self.detect_language(text)

        if "language_code" not in detected:
            return detected

        source_language = detected["language_code"]

        # No translation needed
        if source_language == target_language:
            return {
                "source_language": source_language,
                "target_language": target_language,
                "translation": text,
                "quality_note": "Translation not required."
            }

        # Translation model mapping
        model_map = {
            ("es", "en"):
                "Helsinki-NLP/opus-mt-es-en",

            ("fr", "en"):
                "Helsinki-NLP/opus-mt-fr-en",

            ("de", "en"):
                "Helsinki-NLP/opus-mt-de-en",

            ("en", "es"):
                "Helsinki-NLP/opus-mt-en-es",

            ("en", "fr"):
                "Helsinki-NLP/opus-mt-en-fr",

            ("en", "de"):
                "Helsinki-NLP/opus-mt-en-de"
        }

        pair = (
            source_language,
            target_language
        )

        if pair not in model_map:
            return {
                "status": "unsupported",
                "source_language": source_language,
                "target_language": target_language,
                "message":
                    "Translation model is not configured "
                    "for this language pair."
            }

        model_name = model_map[pair]

        try:
            # Load model only once
            if model_name not in self.translation_models:

                self.translation_models[model_name] = pipeline(
                    "translation",
                    model=model_name
                )

            translator = self.translation_models[
                model_name
            ]

            result = translator(
                text[:2000],
                max_length=512
            )

            translated_text = result[0][
                "translation_text"
            ]

            return {
                "source_language": source_language,
                "target_language": target_language,
                "translation": translated_text,
                "quality_note":
                    "Machine translation should be reviewed "
                    "for cultural meaning and context."
            }

        except Exception as e:
            return {
                "status": "error",
                "message": str(e)
            }


    def analyze_cross_lingual(
        self,
        articles_by_language
    ):
        """
        Compare article collections across languages.

        Expected format:
        {
            "en": ["article 1", "article 2"],
            "es": ["article 1"]
        }
        """

        if not articles_by_language:
            return {
                "status": "error",
                "message":
                    "No multilingual articles provided."
            }

        results = {}

        for language, articles in (
            articles_by_language.items()
        ):

            combined_text = " ".join(
                articles
            ).lower()

            words = re.findall(
                r"\b\w{4,}\b",
                combined_text
            )

            word_counts = Counter(words)

            results[language] = {
                "article_count": len(articles),
                "total_words": len(words),
                "common_terms":
                    word_counts.most_common(10)
            }

        return {
            "languages_compared":
                list(articles_by_language.keys()),

            "comparison": results,

            "insight":
                "Differences in common terms and article "
                "volume may reveal different regional "
                "priorities or perspectives."
        }


    def extract_cultural_context(
        self,
        text,
        source_language
    ):
        """
        Identify simple cultural and regional references.
        """

        if not text:
            return {
                "status": "error",
                "message": "No text provided."
            }

        lower_text = text.lower()

        detected_regions = []

        for region, keywords in (
            self.cultural_keywords.items()
        ):

            for keyword in keywords:

                if keyword.lower() in lower_text:

                    detected_regions.append(
                        region
                    )

                    break

        # Remove duplicates
        detected_regions = list(
            dict.fromkeys(detected_regions)
        )

        return {
            "source_language":
                source_language,

            "detected_regions":
                detected_regions,

            "cultural_context_note":
                "Cultural references should be interpreted "
                "in the social, political, and historical "
                "context of the source language."
        }


# Initialize multilingual processor
multilingual = MultilingualProcessor()

print("🌐 Multilingual processor ready!")
```

```python
import re


class ConversationalInterface:
    """
    Conversational AI for natural language interaction
    with NewsBot 2.0.
    """

    def __init__(self, newsbot_system):
        self.newsbot = newsbot_system

        # Store conversation history
        self.conversation_history = []

        # Simple intent keyword rules
        self.intent_keywords = {
            "search": [
                "find", "show", "search", "news about",
                "articles about"
            ],
            "summarize": [
                "summarize", "summary", "give me a summary"
            ],
            "analyze": [
                "analyze", "sentiment", "trend",
                "patterns", "analysis"
            ],
            "compare": [
                "compare", "difference", "versus", "vs"
            ],
            "explain": [
                "explain", "relationship", "connected",
                "connection", "why"
            ]
        }


    def classify_intent(self, user_query):
        """
        Classify the user's main intent.

        Possible intents:
        - search
        - summarize
        - analyze
        - compare
        - explain
        """

        query = user_query.lower()

        scores = {}

        for intent, keywords in self.intent_keywords.items():

            score = 0

            for keyword in keywords:
                if keyword in query:
                    score += 1

            scores[intent] = score

        # Select highest scoring intent
        best_intent = max(
            scores,
            key=scores.get
        )

        # Default to search if nothing matches
        if scores[best_intent] == 0:
            best_intent = "search"

        return {
            "intent": best_intent,
            "scores": scores
        }


    def extract_query_entities(self, user_query):
        """
        Extract useful information from the query.

        Detects:
        - Sentiment
        - Timeframe
        - Category
        - Possible named entities
        """

        query = user_query.lower()

        entities = {
            "sentiment": None,
            "timeframe": None,
            "category": None,
            "named_entities": []
        }

        # Detect sentiment
        if "positive" in query:
            entities["sentiment"] = "positive"

        elif "negative" in query:
            entities["sentiment"] = "negative"

        elif "neutral" in query:
            entities["sentiment"] = "neutral"


        # Detect timeframe
        if "today" in query:
            entities["timeframe"] = "today"

        elif "this week" in query:
            entities["timeframe"] = "this week"

        elif "last week" in query:
            entities["timeframe"] = "last week"

        elif "this month" in query:
            entities["timeframe"] = "this month"

        elif "last month" in query:
            entities["timeframe"] = "last month"


        # Detect category
        categories = [
            "technology",
            "business",
            "sports",
            "politics",
            "health",
            "science",
            "entertainment",
            "cybersecurity"
        ]

        for category in categories:
            if category in query:
                entities["category"] = category
                break


        # Simple named entity detection
        # Finds capitalized words such as Apple, Google, Tesla
        possible_entities = re.findall(
            r"\b[A-Z][a-zA-Z]+\b",
            user_query
        )

        # Remove common beginning words
        stop_entities = {
            "Show",
            "Find",
            "Compare",
            "Summarize",
            "Analyze",
            "Explain",
            "What",
            "How"
        }

        entities["named_entities"] = [
            item for item in possible_entities
            if item not in stop_entities
        ]

        return entities


    def process_query(
        self,
        user_query,
        conversation_context=None
    ):
        """
        Main conversational processing function.

        Handles:
        - Intent classification
        - Entity extraction
        - Context
        - Response generation
        """

        intent_result = self.classify_intent(
            user_query
        )

        intent = intent_result["intent"]

        entities = self.extract_query_entities(
            user_query
        )

        # Add previous context when available
        if conversation_context:

            for key, value in conversation_context.items():

                if (
                    key in entities
                    and entities[key] is None
                ):
                    entities[key] = value

        # Basic query result
        query_results = {
            "query": user_query,
            "intent": intent,
            "entities": entities
        }

        # Try passing query to NewsBot system
        if self.newsbot is not None:

            try:
                system_result = self.newsbot.process_query(
                    user_query
                )

                query_results["system_result"] = (
                    system_result
                )

            except Exception:
                query_results["system_result"] = None


        response = self.generate_response(
            query_results,
            intent,
            entities
        )

        # Save conversation
        self.conversation_history.append({
            "query": user_query,
            "intent": intent,
            "entities": entities,
            "response": response
        })

        return response


    def generate_response(
        self,
        query_results,
        intent,
        entities
    ):
        """
        Generate a natural language response.
        """

        category = entities.get("category")
        timeframe = entities.get("timeframe")
        sentiment = entities.get("sentiment")
        named_entities = entities.get(
            "named_entities",
            []
        )

        details = []

        if category:
            details.append(
                f"category: {category}"
            )

        if timeframe:
            details.append(
                f"timeframe: {timeframe}"
            )

        if sentiment:
            details.append(
                f"sentiment: {sentiment}"
            )

        if named_entities:
            details.append(
                "entities: "
                + ", ".join(named_entities)
            )

        detail_text = (
            ", ".join(details)
            if details
            else "no specific filters"
        )


        if intent == "search":

            return (
                f"I understand that you want to search "
                f"for news with {detail_text}."
            )


        elif intent == "summarize":

            return (
                f"I understand that you want a summary "
                f"using {detail_text}."
            )


        elif intent == "analyze":

            return (
                f"I understand that you want to analyze "
                f"news content using {detail_text}."
            )


        elif intent == "compare":

            return (
                f"I understand that you want to compare "
                f"news coverage using {detail_text}."
            )


        elif intent == "explain":

            return (
                f"I understand that you want an "
                f"explanation using {detail_text}."
            )


        return "I am ready to help with your news query."


    def handle_follow_up(
        self,
        follow_up_query,
        conversation_history=None
    ):
        """
        Handle follow-up questions using
        previous conversation context.
        """

        history = (
            conversation_history
            if conversation_history is not None
            else self.conversation_history
        )

        if not history:
            return self.process_query(
                follow_up_query
            )

        previous = history[-1]

        previous_entities = previous.get(
            "entities",
            {}
        )

        return self.process_query(
            follow_up_query,
            conversation_context=previous_entities
        )


# Initialize conversational interface
# Ensure newsbot and config are initialized, as they are used here
if 'config' not in globals():
    try:
        # These classes are defined in cell H2zm6ibcquLi
        # If that cell hasn't been run, these will cause a NameError for the classes
        class NewsBot2Config:
            def __init__(self):
                self.news_api_key = None
                self.news_api_endpoint = "https://newsapi.org/v2/everything"
                self.classification_model = "logistic_regression"
                self.transformer_model = "distilbert-base-uncased"
                self.num_topics = 5
                self.data_path = "data/"
                self.model_path = "models/"
                self.output_path = "outputs/"
                self.max_articles = 1000
                self.max_text_length = 5000
                self.confidence_threshold = 0.60
                self.default_language = "en"
                self.supported_languages = ["en", "es", "fr"]

        class NewsBot2System:
            def __init__(self, config):
                self.config = config
                self.preprocessor = None
                self.classifier = None
                self.topic_model = None
                self.language_model = None
                self.multilingual_processor = None
                self.chat_history = []

            def analyze_article(self, article_text):
                if not article_text:
                    return {"status": "error", "message": "No article text provided."}
                analysis = {"status": "success", "character_count": len(article_text), "word_count": len(article_text.split()), "preview": article_text[:200], "classification": None, "topics": None, "sentiment": None}
                return analysis

            def process_query(self, user_query):
                if not user_query:
                    return "Please enter a question."
                self.chat_history.append(user_query)
                response = {"query": user_query, "message": "Query received successfully.", "history_length": len(self.chat_history)}
                return response

            def generate_insights(self, articles):
                if not articles:
                    return {"status": "error", "message": "No articles provided."}
                total_articles = len(articles)
                word_counts = [len(article.split()) for article in articles if isinstance(article, str)]
                insights = {"status": "success", "total_articles": total_articles, "average_word_count": (sum(word_counts) / len(word_counts) if word_counts else 0), "message": "Article collection analyzed successfully."}
                return insights

        config = NewsBot2Config()
        newsbot = NewsBot2System(config)
        print("Initialized NewsBot2Config and NewsBot2System within this cell.")
    except NameError:
        print("Warning: NewsBot2Config or NewsBot2System classes are not defined. Please run cell H2zm6ibcquLi first.")

conversation = ConversationalInterface(newsbot)

print("💬 Conversational interface ready!")
```

```python
# 🔧 System Integration and Orchestration

class NewsBot2IntegratedSystem:
    """
    Complete NewsBot 2.0 system with all major components integrated.
    """

    def __init__(self, config):
        self.config = config

        print("🔧 Initializing NewsBot 2.0 components...")

        # Initialize components safely with NameError handling
        try:
            self.classifier = AdvancedNewsClassifier()
        except NameError:
            self.classifier = None
            print("⚠️ Warning: AdvancedNewsClassifier not defined. Please run its defining cell (WQnb1QdXr9bX).")

        try:
            self.topic_engine = TopicDiscoveryEngine(
                n_topics=getattr(config, "num_topics", 5)
            )
        except NameError:
            self.topic_engine = None
            print("⚠️ Warning: TopicDiscoveryEngine not defined. Please run its defining cell (ve_A4345z7UE).")

        try:
            self.sentiment_tracker = SentimentEvolutionTracker()
        except NameError:
            self.sentiment_tracker = None
            print("⚠️ Warning: SentimentEvolutionTracker not defined. Please run its defining cell (C9EAUagl0Z20).")

        try:
            self.entity_mapper = EntityRelationshipMapper()
        except NameError:
            self.entity_mapper = None
            print("⚠️ Warning: EntityRelationshipMapper not defined. Please run its defining cell (3kn_h0Kq09ss).")

        try:
            self.summarizer = IntelligentSummarizer()
        except NameError:
            self.summarizer = None
            print("⚠️ Warning: IntelligentSummarizer not defined. Please run its defining cell (qdClDkMj3z5Z).")

        try:
            self.search_engine = SemanticSearchEngine()
        except NameError:
            self.search_engine = None
            print("⚠️ Warning: SemanticSearchEngine not defined. Please run its defining cell (MUd4b8qJ4MEy).")

        try:
            self.enhancer = ContentEnhancer()
        except NameError:
            self.enhancer = None
            print("⚠️ Warning: ContentEnhancer not defined. Please run its defining cell (iCSPRK7G4xHe).")

        try:
            self.multilingual = MultilingualProcessor()
        except NameError:
            self.multilingual = None
            print("⚠️ Warning: MultilingualProcessor not defined. Please run its defining cell (J8NHhZpD5Sd8).")

        # Conversational interface connects back to this system
        try:
            self.conversation = ConversationalInterface(self)
        except NameError:
            self.conversation = None
            print("⚠️ Warning: ConversationalInterface not defined. Please run its defining cell (3FqUgmIU7qhV).")

        # Simple system state
        self.analysis_cache = {}
        self.article_database = []

        print("✅ NewsBot 2.0 components initialized (with warnings if any were skipped)!")


    def _safe_run(self, component_attr_name, function, *args, **kwargs):
        """
        Run a component's function safely, checking if the component is initialized.
        """
        component_instance = getattr(self, component_attr_name, None)

        if component_instance is None:
            return {
                "status": "error",
                "component": component_attr_name,
                "message": f"Component '{component_attr_name}' is not initialized. Cannot call function."
            }
        try:
            return function(*args, **kwargs)

        except Exception as error:
            return {
                "status": "error",
                "component": component_attr_name,
                "message": str(error)
            }


    def comprehensive_analysis(self, article_text):
        """
        Perform complete analysis of a single article.
        """

        if not article_text:
            return {
                "status": "error",
                "message": "No article text provided."
            }

        # Return cached result if available
        if article_text in self.analysis_cache:
            return self.analysis_cache[article_text]

        analysis_results = {
            "classification": None,
            "sentiment": None,
            "entities": None,
            "topics": None,
            "summary": None,
            "enhancements": None,
            "language": None
        }

        # Classification
        analysis_results["classification"] = self._safe_run(
            "classifier",
            self.classifier.predict_with_confidence,
            article_text
        )

        # Sentiment
        analysis_results["sentiment"] = self._safe_run(
            "sentiment_tracker",
            self.sentiment_tracker.analyze_sentiment,
            article_text
        )

        # Entity extraction
        analysis_results["entities"] = self._safe_run(
            "entity_mapper",
            self.entity_mapper.extract_entities,
            article_text
        )

        # Topic analysis
        analysis_results["topics"] = self._safe_run(
            "topic_engine",
            self.topic_engine.get_article_topics,
            article_text
        )

        # Summarization
        analysis_results["summary"] = self._safe_run(
            "summarizer",
            self.summarizer.summarize_article,
            article_text,
            "brief"
        )

        # Content enhancement
        analysis_results["enhancements"] = self._safe_run(
            "enhancer",
            self.enhancer.enhance_article,
            article_text
        )

        # Language detection
        analysis_results["language"] = self._safe_run(
            "multilingual",
            self.multilingual.detect_language,
            article_text
        )

        # Save to cache
        self.analysis_cache[article_text] = analysis_results

        return analysis_results


    def batch_analysis(self, articles):
        """
        Analyze multiple articles with progress tracking
        and individual error handling.
        """

        if not articles:
            return {
                "status": "error",
                "message": "No articles provided."
            }

        results = []

        total = len(articles)

        for index, article in enumerate(articles, start=1):

            print(
                f"Analyzing article {index}/{total}..."
            )

            try:
                result = self.comprehensive_analysis(
                    article
                )

                results.append({
                    "article_id": index,
                    "status": "success",
                    "analysis": result
                })

            except Exception as error:

                results.append({
                    "article_id": index,
                    "status": "error",
                    "message": str(error)
                })

        return results


    def process_query(self, user_query):
        """
        Basic system-level query handler.

        This method is used by the conversational interface.
        """

        if not user_query:
            return {
                "status": "error",
                "message": "No query provided."
            }

        return {
            "status": "success",
            "query": user_query,
            "message":
                "NewsBot received the query successfully."
        }


    def query_interface(self, user_query):
        """
        Main entry point for conversational interaction.
        """
        if self.conversation: # Check if conversation interface is initialized
            return self.conversation.process_query(
                user_query
            )
        else:
            return {"status": "error", "message": "Conversational interface not initialized."}


    def generate_insights_report(
        self,
        articles,
        report_type="comprehensive"
    ):
        """
        Generate an insights report from multiple articles.

        Report types:
        - summary
        - comprehensive
        - trends
        - comparative
        """

        if not articles:
            return {
                "status": "error",
                "message": "No articles provided."
            }

        report = {
            "report_type": report_type,
            "article_count": len(articles)
        }

        # High-level content insights
        report["content_insights"] = self._safe_run(
            "enhancer",
            self.enhancer.generate_insights,
            articles
        )

        # Multi-article summary
        report["summary"] = self._safe_run(
            "summarizer",
            self.summarizer.summarize_multiple_articles,
            articles
        )

        # Entity knowledge graph
        report["knowledge_graph"] = self._safe_run(
            "entity_mapper",
            self.entity_mapper.build_knowledge_graph,
            articles
        )

        # Report-specific analysis
        if report_type == "summary":

            return {
                "report_type": "summary",
                "article_count": len(articles),
                "summary": report["summary"]
            }


        elif report_type == "trends":

            if self.entity_mapper:
                report["trending_entities"] = self._safe_run(
                    "entity_mapper",
                    self.entity_mapper.get_top_entities
                )
            else:
                report["trending_entities"] = {"status": "error", "message": "Entity mapper not initialized."}


        elif report_type == "comparative":

            report["comparison_note"] = (
                "Comparative analysis can be extended "
                "using source, language, topic, or sentiment."
            )


        elif report_type == "comprehensive":

            report["batch_analysis"] = (
                self.batch_analysis(articles)
            )

        return report


# Ensure NewsBot2Config is defined.
# If not, provide a minimal placeholder to prevent NameError, and warn the user.
if 'NewsBot2Config' not in globals():
    class NewsBot2Config:
        def __init__(self):
            self.num_topics = 5
    print("⚠️ Warning: NewsBot2Config not found in global scope. Using a minimal placeholder. Please run cell H2zm6ibcquLi first.")

# Initialize complete NewsBot 2.0 system
config = NewsBot2Config()
newsbot2 = NewsBot2IntegratedSystem(config)

print("🔧 Integrated system ready!")
```

```python
# 📊 System Evaluation and Metrics

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

import numpy as np
import pandas as pd
import re


class NewsBot2Evaluator:
    """
    Comprehensive evaluation framework for NewsBot 2.0.
    """

    def __init__(self, newsbot_system):
        self.newsbot = newsbot_system

        # Store evaluation results
        self.results = {}


    def evaluate_classification_performance(self, test_data):
        """
        Evaluate classification performance.

        Expected test_data format:
        DataFrame with columns:
        - text
        - label
        """

        if test_data is None or len(test_data) == 0:
            return {
                "status": "error",
                "message": "No classification test data provided."
            }

        y_true = []
        y_pred = []
        confidences = []

        for _, row in test_data.iterrows():

            result = (
                self.newsbot
                .classifier
                .predict_with_confidence(
                    row["text"]
                )
            )

            # Skip if classifier is not trained
            if (
                isinstance(result, dict)
                and result.get("status") == "error"
            ):
                continue

            y_true.append(
                row["label"]
            )

            y_pred.append(
                result["primary_category"]
            )

            confidences.append(
                result["confidence"]
            )

        if len(y_true) == 0:
            return {
                "status": "skipped",
                "message":
                    "Classifier must be trained before evaluation."
            }

        accuracy = accuracy_score(
            y_true,
            y_pred
        )

        precision = precision_score(
            y_true,
            y_pred,
            average="weighted",
            zero_division=0
        )

        recall = recall_score(
            y_true,
            y_pred,
            average="weighted",
            zero_division=0
        )

        f1 = f1_score(
            y_true,
            y_pred,
            average="weighted",
            zero_division=0
        )

        matrix = confusion_matrix(
            y_true,
            y_pred
        )

        report = classification_report(
            y_true,
            y_pred,
            output_dict=True,
            zero_division=0
        )

        result = {
            "accuracy": round(accuracy, 4),
            "precision": round(precision, 4),
            "recall": round(recall, 4),
            "f1_score": round(f1, 4),
            "average_confidence": round(
                float(np.mean(confidences)),
                4
            ),
            "confusion_matrix": matrix.tolist(),
            "classification_report": report
        }

        self.results[
            "classification"
        ] = result

        return result


    def evaluate_topic_modeling_quality(
        self,
        documents
    ):
        """
        Evaluate topic modeling quality.

        Uses simple measures:
        - Topic diversity
        - Vocabulary overlap
        - Interpretability proxy
        """

        if not documents:
            return {
                "status": "error",
                "message": "No documents provided."
            }

        try:

            topics = (
                self.newsbot
                .topic_engine
                .get_top_words(
                    top_n=10
                )
            )

            if isinstance(topics, str):
                return {
                    "status": "skipped",
                    "message": topics
                }

            all_words = []

            for words in topics.values():
                all_words.extend(words)

            unique_words = set(
                all_words
            )

            topic_diversity = (
                len(unique_words)
                / len(all_words)
                if all_words
                else 0
            )

            # Average number of unique words per topic
            avg_unique_words = np.mean([
                len(set(words))
                for words in topics.values()
            ])

            result = {
                "number_of_topics": len(topics),
                "topic_diversity": round(
                    topic_diversity,
                    4
                ),
                "average_unique_words_per_topic":
                    round(
                        float(avg_unique_words),
                        2
                    ),
                "topics": topics,
                "interpretability_note":
                    "Higher topic diversity generally "
                    "indicates less overlap between topics."
            }

            self.results[
                "topic_modeling"
            ] = result

            return result

        except Exception as e:
            return {
                "status": "error",
                "message": str(e)
            }


    def evaluate_summarization_quality(
        self,
        articles_and_summaries
    ):
        """
        Evaluate summarization quality.

        Expected format:
        [
            {
                "article": "...",
                "summary": "..."
            }
        ]
        """

        if not articles_and_summaries:
            return {
                "status": "error",
                "message":
                    "No article-summary pairs provided."
            }

        evaluations = []

        for item in articles_and_summaries:

            article = item["article"]
            summary = item["summary"]

            quality = (
                self.newsbot
                .summarizer
                .assess_summary_quality(
                    article,
                    summary
                )
            )

            evaluations.append(
                quality
            )

        average_compression = np.mean([
            item.get(
                "compression_ratio",
                0
            )
            for item in evaluations
        ])

        average_coverage = np.mean([
            item.get(
                "information_coverage",
                0
            )
            for item in evaluations
        ])

        result = {
            "samples_evaluated":
                len(evaluations),

            "average_compression_ratio":
                round(
                    float(
                        average_compression
                    ),
                    4
                ),

            "average_information_coverage":
                round(
                    float(
                        average_coverage
                    ),
                    4
                ),

            "individual_results":
                evaluations
        }

        self.results[
            "summarization"
        ] = result

        return result


    def evaluate_user_experience(
        self,
        user_interactions
    ):
        """
        Evaluate conversational interface effectiveness.

        Expected format:
        [
            {
                "query": "...",
                "expected_intent": "search",
                "satisfaction": 5,
                "task_completed": True
            }
        ]
        """

        if not user_interactions:
            return {
                "status": "error",
                "message":
                    "No user interactions provided."
            }

        correct_intents = 0
        satisfaction_scores = []
        completed_tasks = 0

        detailed_results = []

        for interaction in user_interactions:

            query = interaction[
                "query"
            ]

            expected_intent = (
                interaction.get(
                    "expected_intent"
                )
            )

            predicted = (
                self.newsbot
                .conversation
                .classify_intent(
                    query
                )
            )

            predicted_intent = (
                predicted["intent"]
            )

            intent_correct = (
                predicted_intent
                == expected_intent
            )

            if intent_correct:
                correct_intents += 1

            satisfaction = (
                interaction.get(
                    "satisfaction"
                )
            )

            if satisfaction is not None:
                satisfaction_scores.append(
                    satisfaction
                )

            if interaction.get(
                "task_completed",
                False
            ):
                completed_tasks += 1

            detailed_results.append({
                "query": query,
                "expected_intent":
                    expected_intent,
                "predicted_intent":
                    predicted_intent,
                "correct":
                    intent_correct
            })

        total = len(
            user_interactions
        )

        intent_accuracy = (
            correct_intents / total
        )

        task_completion_rate = (
            completed_tasks / total
        )

        avg_satisfaction = (
            np.mean(
                satisfaction_scores
            )
            if satisfaction_scores
            else None
        )

        result = {
            "query_understanding_accuracy":
                round(
                    intent_accuracy,
                    4
                ),

            "task_completion_rate":
                round(
                    task_completion_rate,
                    4
                ),

            "average_user_satisfaction":
                (
                    round(
                        float(
                            avg_satisfaction
                        ),
                        2
                    )
                    if avg_satisfaction
                    is not None
                    else None
                ),

            "detailed_results":
                detailed_results
        }

        self.results[
            "user_experience"
        ] = result

        return result


    def generate_evaluation_report(self):
        """
        Generate a combined evaluation report.
        """

        if not self.results:
            return {
                "status": "warning",
                "message":
                    "Run evaluation methods before generating a report."
            }

        report = {
            "system":
                "NewsBot 2.0",

            "components_evaluated":
                list(
                    self.results.keys()
                ),

            "results":
                self.results,

            "strengths": [
                "Supports multiple NLP tasks",
                "Uses semantic search and transformer models",
                "Includes multilingual processing",
                "Supports conversational interaction",
                "Includes error handling and system testing"
            ],

            "limitations": [
                "Some models require training before evaluation",
                "Transformer models can require significant memory",
                "Translation support is limited to configured languages",
                "Fact-checking currently identifies claims but does not verify them automatically"
            ],

            "recommendations": [
                "Train classification and topic models on larger datasets",
                "Add human evaluation for summaries and responses",
                "Expand multilingual support",
                "Improve factual verification using trusted external sources",
                "Optimize large models for faster processing"
            ]
        }

        return report


# Initialize evaluator
evaluator = NewsBot2Evaluator(
    newsbot2
)

print("📊 Evaluation framework ready!")
```
